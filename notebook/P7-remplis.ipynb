```python
import pandas as pd
import numpy as np
from datetime import datetime
import matplotlib.pyplot as plt
```

# Charger les DF des différents pages



```python
# Chargement des données
projects_plans = pd.read_excel(FN, sheet_name='Projects_plans', header=2)
actual_duration = pd.read_excel(FN, sheet_name='Actual_Duration', header=5)
project_type = pd.read_excel(FN, sheet_name='Project type', header=3)
locations = pd.read_excel(FN, sheet_name='Projects_Locations', header=1)
country_profiles = pd.read_excel(FN, sheet_name='Country_Profiles', header=1)
```

```python
projects_plans.dtypes
```

```python
actual_duration
```

```python
df = pd.merge(projects_plans, actual_duration, 
              left_on=['Project ID', 'Phase'], 
              right_on=['Project', 'Phase'], 
              how='outer', 
              indicator = True)
df
```

```python
df._merge.value_counts()
```

```python
# Calcul du retard (en jours)
df['Delay'] = df['Actual_Duration'] - df['Planned_Duration']
df['Is_Late'] = (df['Delay'] > 0).astype(int)  # Variable binaire : 1 = retard, 0 = pas de retard

# Degré de retard (catégories)
df['Delay_Category'] = pd.cut(df['Delay'], 
                               bins=[-float('inf'), -1, 0, 30, 90, float('inf')],
                               labels=['En avance', 'À l\'heure', 'Petit retard (<1mois)', 
                                      'Retard moyen (1-3mois)', 'Grand retard (>3mois)'])
df
```

```python
# Mois de début (saisonnalité)
df['Start_Month'] = pd.to_datetime(df['Start Date']).dt.month
df['Start_Quarter'] = pd.to_datetime(df['Start Date']).dt.quarter
df['Start_Year'] = pd.to_datetime(df['Start Date']).dt.year

# Charge de travail initiale
df['Planned_Intensity'] = df['Planned_Delivrable'] / df['Planned_Duration'] * 30  # livrables/mois
df
```

```python
project_features = df.groupby('Project ID').agg({
    'Planned_Duration': 'sum',
    'Planned_Cost': 'sum',
    'Planned_Delivrable': 'sum',
    'Phase': 'count',
    'Delay': 'sum',
    'Is_Late': 'max',
    'Start_Month': 'first',
    'Start_Year': 'first'
}).rename(columns={
    'Planned_Duration': 'Total_Planned_Duration',
    'Planned_Cost': 'Total_Planned_Cost',
    'Planned_Delivrable': 'Total_Planned_Delivrables',
    'Phase': 'Nb_Phases',
    'Delay': 'Total_Delay',
    'Is_Late': 'Project_Had_Late_Phase'
})

project_features
```

```python
# Fusion avec le type de projet
df = pd.merge(df, project_type, on='Project ID', how='left')

# Fusion avec la localisation
df = pd.merge(df, locations, on='Project ID', how='left')
df = pd.merge(df, country_profiles, on='Country', how='left')

df
```
